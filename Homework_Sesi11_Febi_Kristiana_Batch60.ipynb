```python
def terbesar(x, y, z):
    if x >= y and x >= z:
        return x
    elif y >= x and y >= z:
        return y
    else:
        return z

# Contoh penggunaan
print(terbesar(50, 40, 60))  # Output: 60
```

2. Buatlah sebuah fungsi untuk mendapatkan nilai rata-rata dari input sejumlah n  bilangan dalam sebuah list. Catatan: tidak boleh menggunakan fungsisum(), mean()  atau built-in function lainnya.
Contoh input: [10, 20, 30, 40, 50]  
Output yang diharapkan: 30


```python
def rata_rata(lst):
    total = 0
    count = 0
    for num in lst:
        total += num
        count += 1
    return total // count  # Menggunakan pembagian bulat agar hasilnya berupa integer

# Contoh penggunaan
print(rata_rata([10, 20, 30, 40, 50]))  # Output: 30
print(rata_rata([10, 15, 20, 25, 30, 35, 40, 45, 50]))  # Output: 30
```

Fungsi ini menerima sebuah list berisi angka-angka dan menghitung rata-rata dari angka-angka tersebut.
Kita menggunakan dua variabel: satu untuk menjumlahkan semua angka (total) dan satu lagi untuk menghitung jumlah angka yang ada (count).
Setelah itu, kita menghitung rata-rata dengan cara membagi total jumlah angka dengan banyaknya angka dalam list.

3. Buatlah sebuah fungsi yang menghitung banyaknya bilangan genap dan bilangan  ganjil dari sejumlah elemen di dalam sebuah list. Output berupa sebuah tuple  dengan dua elemen: elemen pertama adalah banyanya bilangan genap, elemen  kedua adalah banyaknya bilangan ganjil.
Contoh input: [1, 2, 3, 4, 5, 6, 7]
Output yang diharapkan: (3, 4)


```python
def hitung_genap_ganjil(lst):
    genap = 0
    ganjil = 0
    for num in lst:
        if num % 2 == 0:
            genap += 1
        else:
            ganjil += 1
    return (genap, ganjil)

# Contoh penggunaan
print(hitung_genap_ganjil([1, 2, 3, 4, 5, 6, 7]))  # Output: (3, 4)
print(hitung_genap_ganjil([11, 12, 13, 14, 15, 16, 17]))  # Output: (3, 4)
```

Fungsi ini akan menerima sebuah list yang berisi angka-angka dan menghitung berapa banyak bilangan genap dan berapa banyak bilangan ganjil.
Kita menggunakan operator modulus % untuk memeriksa apakah angka tersebut genap atau ganjil. Jika angka tersebut habis dibagi 2, maka itu adalah bilangan genap.

4. Diberikan sebuah string yang merupakan campuran dari huruf kecil dan huruf  kapital. Buatlah sebuah fungsi yang mengubah susunan huruf dalam string tersebut  supaya huruf kecil muncul terlebih dahulu baru diikuti huruf kapital dengan tetap  mempertahankan urutan kemunculan karakter aslinya.
Contoh input: “DigitaLSkolA”
Contoh output: “igitakolDLSA


```python
def urut_huruf(string):
    kecil = ""
    kapital = ""
    for char in string:
        if char.islower():
            kecil += char
        else:
            kapital += char
    return kecil + kapital

# Contoh penggunaan
print(urut_huruf("DigitaLSkolA"))  # Output: "igitakolDLSA"
print(urut_huruf("FebiKristianA"))  # Output: "eviristianFKA"
```

Fungsi ini akan memisahkan huruf kecil dan huruf kapital dalam string yang diberikan, kemudian menggabungkannya dengan huruf kecil di depan dan huruf kapital di belakang.
Fungsi ini memeriksa apakah karakter dalam string adalah huruf kecil atau kapital, dan menyimpannya dalam dua variabel terpisah.

5. Diberikan suatu kata dalam string, buatlah fungsi yang menghitung banyaknya  kemunculan setiap karakter (huruf/angka/symbol) dalam kata tersebut. Hasil  kembaliannya harus berupa dictionary. Dilarangmenggunakan fungsi built-in count()  dari Python.
Contoh input: “data science”
Contoh output: {'d': 1, 'a': 2, 't': 1, ' ': 1, 's': 1, 'c': 2, 'i': 1, 'e': 2, 'n': 1}


```python
def hitung_kemunculan(kata):
    hasil = {}
    for char in kata:
        if char in hasil:
            hasil[char] += 1
        else:
            hasil[char] = 1
    return hasil

# Contoh penggunaan
print(hitung_kemunculan("data science"))  # Output: {'d': 1, 'a': 2, 't': 1, ' ': 1, 's': 1, 'c': 2, 'i': 1, 'e': 2, 'n': 1}
print(hitung_kemunculan("febi123"))  # Output: {'f': 1, 'e': 1, 'b': 1, 'i': 1, '1': 1, '2': 1, '3':
```

Fungsi ini menerima sebuah string dan menghitung berapa kali setiap karakter (huruf, angka, atau simbol) muncul dalam string tersebut.
Kita menggunakan dictionary untuk menyimpan hasilnya, di mana key adalah karakter dan value adalah jumlah kemunculannya.

6. Buatlah sebuah fungsi dengan nama show_employee() berdasarkan ketentuan  berikut:
• Fungsi harus menerima nama dan gaji employee dan menampilkan keduanya di  layar.
• Jika argumen gaji tidak diisi/dikosongkan saat pemanggilan fungsi, otomatis  gunakan nilai 2.000.000
• Fungsi juga mengembalikan suatu dictionary dengan key-nya adalah “nama”  dan “gaji”, value-nya adalah nilai nama dan gajinya.

```python
def show_employee(nama, gaji=2000000):
    print(f"Nama: {nama}, Gaji: {gaji}")
    return {"nama": nama, "gaji": gaji}

# Contoh penggunaan
print(show_employee("Ayunda Risu"))  # Output: Nama: Ayunda Risu, Gaji: 2000000, {'nama': 'Ayunda Risu', 'gaji': 2000000}
print(show_employee("Kobo Kanaeru", 2500000))  # Output: Nama: Kobo Kanaeru, Gaji: 2500000, {'nama': 'Kobo Kanaeru', 'gaji': 2500000}
print(show_employee("Vestia Zeta", 5500000))  # Output: Nama: Vestia Zeta, Gaji: 2500000, {'nama': 'Vestia Zeta', 'gaji': 5500000}
```

Fungsi ini menerima dua argumen: nama dan gaji. Jika argumen gaji tidak diberikan, maka akan digunakan nilai default yaitu 2.000.000.
Fungsi ini akan menampilkan informasi nama dan gaji, dan juga mengembalikan informasi tersebut dalam bentuk dictionary.
